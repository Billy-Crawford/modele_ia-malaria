```python
import os 
import random
import matplotlib.pyplot as plt
from PIL import Image

BASE_DIR = "../data/raw/cell_images"
classes = ["Parasitized", "Uninfected"]

plt.figure(figsize=(10, 5))

for i, cls in enumerate(classes):
    folder = os.path.join(BASE_DIR, cls)
    img_name = random.choice(os.listdir(folder))
    img_path = os.path.join(folder, img_name)

    img = Image.open(img_path)

    plt.subplot(1, 2, i + 1)
    plt.imshow(img)
    plt.title(cls)
    plt.axis("off")

plt.show()
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

# Chemin du dataset brut
RAW_DIR = "../data/raw/cell_images"
# Chemin des splits
SPLIT_DIR = "../data/splits"

# Créer les dossiers si n'existent pas
for split in ["train", "val", "test"]:
    for cls in ["Parasitized", "Uninfected"]:
        os.makedirs(os.path.join(SPLIT_DIR, split, cls), exist_ok=True)

# Pour chaque classe, créer les splits
for cls in ["Parasitized", "Uninfected"]:
    cls_dir = os.path.join(RAW_DIR, cls)
    images = os.listdir(cls_dir)

    train_and_val, test = train_test_split(images, test_size=0.15, random_state=42)
    train, val = train_test_split(train_and_val, test_size=0.1765, random_state=42) # 0.1765*0.85≈0.15   

    # copier les images
    for img in train:
        shutil.copy(os.path.join(cls_dir, img), os.path.join(SPLIT_DIR, "train", cls, img))
    for img in val:
        shutil.copy(os.path.join(cls_dir, img), os.path.join(SPLIT_DIR, "val", cls, img))
    for img in test:
        shutil.copy(os.path.join(cls_dir, img), os.path.join(SPLIT_DIR, "test", cls, img))

print("Splits crees avec succes !")
```
